### Export workflow overview

Every chart in this notebook can be saved to PNG / SVG / HTML with a single
`status.export.save(chart, 'descriptive-name')` call. The output directory,
file-naming convention, target formats and raster scale factor are all
configured once in the cell above and visible below.

```python
src_file = RAW_DATA_DIR / 'AmazonBooks.xlsx'
df = read_excel_safe(src_file)
df.head()
```

```python
yearly_reviews = alt.Chart(df).mark_bar().encode(
    alt.Y('Year:O', title='Published Year'),
    alt.X('sum(Reviews)', title='Number of Reviews'),
    color='Genre'
)
yearly_reviews
```

```python
status.export.save(yearly_reviews, 'amazon-yearly-reviews-by-genre')
```

```python
alt.Chart(df).mark_rect().encode(
    x='Year:O',
    y='Genre:O',
    color='mean(Price):Q',
    tooltip=[alt.Tooltip('mean(Price):Q', format='$.2f'), 
             alt.Tooltip('count(Name):Q', format='.0f')]
)
```

```python
df.info()
```

```python
df['Author'].value_counts()
```

```python
top_authors = list(
    df.groupby(["Author"], as_index=False)
    .agg({"Reviews": "sum"})
    .nlargest(20, columns=["Reviews"])["Author"]
)
```

```python
top_authors
```

```python
alt.Chart(df.query('Author == @top_authors')).mark_circle(opacity=0.8,
    stroke='black',
    strokeWidth=1
).encode(
    alt.Y('Author'),
    alt.X('Year:O'),
    alt.Size('sum(Reviews)', 
             scale=alt.Scale(range=[0,500]),
             legend=alt.Legend(title='Reviews')),
    alt.Color('Author'))
```

```python
alt.Chart(df.query('Author == @top_authors')).mark_circle(opacity=0.8,
    stroke='black',
    strokeWidth=1
).encode(
    alt.Y('Author'),
    alt.X('Year:O'),
    alt.Size('sum(Reviews)', 
             scale=alt.Scale(range=[0,500]),
             legend=alt.Legend(title='Reviews')),
    alt.Color('Author', legend=None))
```

```python
author_reviews = alt.Chart(df).mark_circle(opacity=0.8,
    stroke='black',
    strokeWidth=1
).encode(
    alt.Y('Author'),
    alt.X('Year:O'),
    alt.Size('sum(Reviews)', 
             scale=alt.Scale(range=[0,900]),
             legend=alt.Legend(title='Reviews')),
    alt.Color('Author', legend=None)
).configure_axis(
    grid=True
).transform_filter(
    alt.FieldOneOfPredicate(field='Author', 
                            oneOf=top_authors)
).properties(
    width=550,
    height=475,
    title='Amazon Author Reviews')
author_reviews
```

```python
status.export.save(author_reviews, 'amazon-top-author-reviews')
```
