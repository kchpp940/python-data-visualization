```python
src_file = datasets.path('epa_summary')
image_dir = IMAGES_DIR
```

```python
df = pd.read_csv(src_file)
df.head()
```

```python
alt.data_transformers.names()
```

```python
alt.Chart(df).mark_circle().encode(
    x='displ',
    y='fuelCost08',
)
```

```python
alt.Chart(df).mark_point().encode(
    x='displ',
    y='fuelCost08',
)
```

```python
alt.Chart(df).mark_point().encode(
    x='displ',
    y='fuelCost08',
    color='drive',
    shape='drive',
)
```

```python
alt.Chart(df).mark_bar().encode(
    x='fuelCost08',
    y='count()',
)
```

```python
alt.Chart(df).mark_tick().encode(
    y='fuel_type_summary',
    x='barrels08'
)
```

```python
alt.Chart(df).mark_boxplot().encode(
    x='year',
    y='fuelCost08'
)
```

```python
alt.Chart(df).mark_boxplot().encode(
    x='year:O',
    y='fuelCost08:Q'
)
```

```python
alt.Chart(df).mark_bar().encode(
    x='mean(fuelCost08)',
    y='year'
)
```

```python
alt.Chart(df).mark_bar().encode(
    x='mean(fuelCost08)',
    y='year:O'
)
```

```python
alt.Chart(df).mark_bar().encode(
    x='fuelCost08',
    y='count()',
)
```

```python
alt.Chart(df).mark_bar().encode(
    alt.X('fuelCost08', type='quantitative', bin=True),
    alt.Y(aggregate='count', type='quantitative'),
)
```

```python
alt.Chart(df).mark_bar().encode(
    alt.X('fuelCost08:Q', bin=alt.Bin(extent=[0,5000], step=250)),
    alt.Y('count()'),
)
```

```python
alt.Chart(df).mark_point().encode(
    alt.X('displ', type='quantitative'),
    alt.Y('fuelCost08'),
    alt.Color('cylinders', type='ordinal')
)
```

```python
alt.Chart(df).mark_point().encode(
    alt.X('displ', type='quantitative'),
    alt.Y('fuelCost08'),
    alt.Color('cylinders', type='quantitative')
)
```
